```python
import random

number= random.randint(1,100)

print("Welcome to the Guess the Number game!")
print("I'm thinking of a number between 1 and 100. Can you guess what it is?")
while True:
  guess = int(input("Enter your guess: "))
  if guess < number
  print("Your guess is too low. Guess again")
  elif guess > number
  print("Your guess is too high. Guess again")
  else:
    print("Congratulations! You guessed the number correctly!")
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_excel("/ML_Student_Placement_Dataset.xlsx")
print(df)
```

```python
df.shape
df.tail()
df.info()
df.describe()
df.isnull().sum()
```

```python
# sns.scatterplot(df['CGPA'],df['Resume Score'],hue=df['Placed'])
# sns.scatterplot(df['CGPA'])
sns.scatterplot(x='CGPA',y='Resume Score',hue='Placed',data=df)
plt.title("CGPA vs Resume Score")
plt.xlabel("CGPA")
plt.ylabel("Resume Score")
plt.show()
```

```python
x=df.iloc[:,:-1]
y=df.iloc[:,-1]
print(x)
print(y)
```
